```python
X = df.drop("selling_price", axis = 1)
y = df["selling_price"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 15)
```

```python
categoric_cols = df.select_dtypes("object").columns.to_list()
```

```python
categoric_cols
```

```python
unique_values = df[categoric_cols].nunique()
```

```python
unique_values
```

```python
# seller, fuel, transmission type -> One-Hot Encoding
# car_name, brand, model -> frequency encoding
```

```python
onehot_cols = ["seller_type","fuel_type","transmission_type"]
freq_cols = ["car_name", "brand", "model"]
```

```python
X_train["car_name"].value_counts()/len(X_train)
```

```python
for col in freq_cols:
    freq = X_train[col].value_counts() / len(X_train)

    X_train[col + "_freq"] = X_train[col].map(freq)

    X_test[col + "_freq"] = X_test[col].map(freq)
    
    mean_freq = freq.mean()
    X_test[col + "_freq"] = X_test[col + "_freq"].fillna(mean_freq)
```

```python
X_test.head()
```

```python
X_train = X_train.drop(["car_name", "brand","model"], axis = 1)
X_test = X_test.drop(["car_name", "brand","model"], axis = 1)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(
    transformers = [
        ("onehot", OneHotEncoder(drop="first", handle_unknown="ignore"), onehot_cols)
    ], remainder = "passthrough"
)
```

```python
X_train = transformer.fit_transform(X_train)
X_test = transformer.transform(X_test)
```

```python
encoded_cols = transformer.get_feature_names_out()
encoded_cols
```

```python
X_train
```

```python
X_train = pd.DataFrame(X_train, columns = encoded_cols)
X_test = pd.DataFrame(X_test, columns = encoded_cols)
```

```python
X_train.head()
```

```python
from sklearn.ensemble import AdaBoostRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = AdaBoostRegressor()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print("r2 score:" ,r2_score(y_pred, y_test))
print("mae:" ,mean_absolute_error(y_pred, y_test))
print("mse:" ,mean_squared_error(y_pred, y_test))
```

```python
params = {
    "n_estimators": [50,70,100,150],
    "learning_rate": [0.0001, 0.01, 0.1, 1, 2],
    "loss": ["square", "linear", "exponential"]
}
```

```python
rcv = RandomizedSearchCV(estimator = AdaBoostRegressor(), param_distributions = params, scoring = "r2", cv = 5)
```

```python
rcv.fit(X_train, y_train)
```

```python
rcv.best_params_
```

```python
y_pred = rcv.predict(X_test)
```

```python
print("r2 score:" ,r2_score(y_pred, y_test))
print("mae:" ,mean_absolute_error(y_pred, y_test))
print("mse:" ,mean_squared_error(y_pred, y_test))
```

```python
# AdaBoost içerisindeki DecisionTree parametrelerini de değiştirerek hyperparameter tuning yapıyoruz
```

```python
params = {
    "estimator__max_depth": [3,4,5,8],
    "n_estimators": [50,70,100,150],
    "learning_rate": [0.0001, 0.01, 0.1, 1, 2],
    "loss": ["square", "linear", "exponential"]
}
```

```python
rcv = RandomizedSearchCV(estimator = AdaBoostRegressor(DecisionTreeRegressor()), param_distributions = params, scoring = "r2", cv = 5, n_jobs = -1)
```

```python
rcv.fit(X_train, y_train)
```

```python
rcv.best_params_
```

```python
y_pred = rcv.predict(X_test)
```

```python
print("r2 score:" ,r2_score(y_pred, y_test))
print("mae:" ,mean_absolute_error(y_pred, y_test))
print("mse:" ,mean_squared_error(y_pred, y_test))
```
