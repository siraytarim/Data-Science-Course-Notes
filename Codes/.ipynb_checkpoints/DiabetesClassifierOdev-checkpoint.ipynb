```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("diabetes.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
cols_for_fill = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]

for col in cols_for_fill:
   df[col] = df[col].replace(0, np.nan)
```

```python
for col in cols_for_fill:
    df[col] = df[col].fillna(df.groupby("Outcome")[col].transform("median"))
```

```python
df[cols_for_fill].isnull().sum()
```

```python
df.describe()
```

```python
X = df.drop("Outcome", axis = 1)
y =df["Outcome"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.15, random_state = 15)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
```

```python
dt_model = DecisionTreeClassifier()
dt_model.fit(X_train, y_train)
y_pred = dt_model.predict(X_test)
```

```python
print("classification_report: \n ", classification_report(y_test, y_pred))
print("accuracy_score: ", accuracy_score(y_test, y_pred))
print("confusion_matrix: \n", confusion_matrix(y_test, y_pred))
```

```python
params = {
    "criterion": ["gini", "entropy", "log_loss"],
    "splitter": ["best", "random"],
    "max_depth":[3,5,7, None],
    "max_features": ["sqrt", "log2", 3, 5, 8]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid = GridSearchCV(estimator = DecisionTreeClassifier(), param_grid = params, cv = 3,verbose = 1, n_jobs = -1)
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
dt_model2 = DecisionTreeClassifier(criterion="entropy", max_depth= 7, max_features= 'sqrt', splitter= 'best')
```

```python
dt_model2.fit(X_train, y_train)
y_pred2 = dt_model2.predict(X_test)
print("classification_report: \n ", classification_report(y_test, y_pred2))
print("accuracy_score: ", accuracy_score(y_test, y_pred2))
print("confusion_matrix: \n", confusion_matrix(y_test, y_pred2))
```

```python
import joblib
joblib.dump(dt_model, 'diabetes_model.pkl')
```
