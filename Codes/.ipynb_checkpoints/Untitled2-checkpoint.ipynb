```python
import pandas as pd
```

```python
data = {
    "yas": [15,20,22,24,28],
    "gelir": [20000, 28000, 35000, 450000, 66000],
    "cinsiyet": ["K", "E", "E", "E","K"],
    "id_no": [120,106,204,145,160]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df_new = df.drop("id_no", axis = 1) # burada inplace = True ile df değişkeni None değerini aldı
df_new
```

```python
df_changed = df.drop(["yas","gelir"], axis = 1)
```

```python
df_changed
```

```python
df
```

```python
silinecekler = ["id_no", "cinsiyet"]
df_son = df.drop(silinecekler, axis = 1)
```

```python
df_son
```

```python
df
```

```python
for col in df.columns:
    if col.endswith("no"):
        df.drop(col, axis = 1, inplace = True)
```

```python
df
```

```python
silinecekler = [col for col in df.columns if col.endswith("t")]
df = df.drop(silinecekler, axis = 1)
```
